# ItemiSet

> Itemized area-proportional Euler diagrams: every item gets one cell, so area is count and the names are the picture.

A Venn diagram tells you how many proteins three baits share, but not which ones. ItemiSet draws the members themselves: one cell per item, coloured by the sets it belongs to, arranged so that each set and each overlap is a single connected region. Area is exactly proportional to count, and the figure doubles as the lookup table.

- **Website and web app:** <https://adamcc.github.io/itemiset/>
- **Documentation:** start with the [tour](https://adamcc.github.io/itemiset/tour.html)

ItemiSet draws one, two or three sets. It is pre-release software (version 0.0.1).

## Install

```sh
pip install git+https://github.com/adamcc/itemiset
```

Add `openpyxl` (or install `itemiset[excel]`) to read Excel files.

## Use

From Python:

```python
from itemiset import plot, load_sets

sets, categories, colours = load_sets("hits.csv")   # or a dict: {"DDX6": [...], "LSM14A": [...]}
plot(sets, out="hits.pdf", categories=categories, category_colours=colours)
```

From the command line:

```sh
itemiset hits.csv -o hits.pdf --zones-out zones.csv
```

The input can be a membership table (a `gene` column plus a yes/no column per set) or one column per set. CSV, TSV and Excel all work.

## Two implementations, one layout

The Python package is the reference. The website runs a JavaScript port (`js/itemiset.js`) that must place every item in the same cell: `nbs/04_fixtures.ipynb` records a few hundred Python layouts, and the JavaScript tests reproduce them exactly.

## Example data

The examples are high-confidence BioID hits from Youn J-Y *et al.* (2018) *Molecular Cell* 69:517–532 ([doi:10.1016/j.molcel.2017.12.020](https://doi.org/10.1016/j.molcel.2017.12.020)), taken from BioGRID release 5.0.262 (Stark *et al.* 2006, *Nucleic Acids Research* 34:D535). BioGRID data are released under the MIT License. See `data/README.md`.

## Development

The Python code is written in the notebooks in `nbs/` with [nbdev](https://nbdev.fast.ai):

```sh
pip install -e ".[dev]" nbdev
nbdev-install-hooks      # once: cleans notebooks on commit
nbdev-export             # notebooks → itemiset/*.py
nbdev-test               # run every notebook as a test
cd js && npm test        # JavaScript layout must match the Python fixtures
```

To build the whole website locally (documentation plus the splash page and web app):

```sh
nbdev-docs && python tools/build_site.py && python -m http.server -d _docs
```
